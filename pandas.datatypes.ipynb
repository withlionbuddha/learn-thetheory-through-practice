| 분류              | 타입       | 설명          | 예시                               |
| ---------------- | -----------| ------------ | -------------------------------- |
| 숫자형 (Numeric)  | int        | 정수          | x = 10                           |
|                  | float      | 실수 (소수점)  | y = 3.14                        |
|                  | complex    | 복소수         | z = 2 + 3j                     |
| 문자형 (Text)     | str        | 문자열         | s = "Hello"                    |
| 불리언 (Boolean)  | bool       | 참/거짓        | flag = True                    |
| 시퀀스 (Sequence) | list       | 순서O, 변경O   | fruits = ["apple", "banana"]   |
|                  | tuple      | 순서O, 변경X   | point = (10, 20)               |
|                  | range      | 숫자 범위      | r = range(5) → [0,1,2,3,4]   |
| 집합형 (Set)      | set        | 순서X, 중복X   | s = {1, 2, 3}                  |
|                  | frozenset  | 변경 불가 집합  | fs = frozenset([1,2,3])        |
| 매핑형 (Mapping)  | dict       | 키:값 쌍       | d = {"name": "Tom", "age": 20} |
| 바이너리 (Binary)  | bytes      | 불변 바이트 시퀀스 | b = b"hi"                  |
|                  | bytearray  | 가변 바이트 시퀀스 | ba = bytearray(5)            |
|                  | memoryview | 버퍼 참조       | mv = memoryview(b"hi")        |

```python
dict_ex1 = {"name": "tom", "age" : 20}
dict_ex2 = {"race": 1, "gend": 1}

# dict 꺼내기 = 속성 접근
dict_ex1["name"] , dict_ex1.get("name")
```

```python
tuple_ex = (dict_ex1, dict_ex2)

tuple_ex[0].get("name")
```
