```python
import pandas as pd

# 예제 DataFrame: 같은 열에 다른 타입의 값들이 있음
df = pd.DataFrame({
    'A': [1, 2, '3'],  # 첫 번째와 두 번째 행은 1, 2 (정수), 세 번째 행은 '3' (문자열)
    'B': [10.0, 20.0, 30.0]
})

# DataFrame 전체에서는 'A' 열의 dtype은 object
print("DataFrame dtypes:")
print(df.dtypes)
```

iterrows()는 각 row별로 seriese 객체를 생성할때, column별로 type이 다르면 공통된 혹은 더 일반적인 타입(object, float 등)으로 업캐스팅되어 type을 반환한다. 즉 원본 DataFrame의 column 단위 타입을 보존하지 않으며, dtype이 강제 캐스팅되면서 원래 데이터 타입 손실이 발생하여 타입 안정성(type safety)문제가 발생될 수 있습니다. 

```python
# iterrows를 사용하여 각 행을 확인
for idx, row in df.iterrows():
    print(f"\nRow {idx}:")
    print(row)
    print("Type of row['A']:", type(row['A']))
```

itertuples()는 각 row별로 객체를 반환할때 type('Pandas', (tuple,), {...}) 으로 구성된 <class 'pandas.core.frame.Pandas'> 객체를 반환한다. 즉 column의 값을 불변(immutable) tuple로 구성된 pandas 객체를 생성하여 반환한다.

```python
# itertuples 사용하여 각 행을 확인
for row in df.itertuples():
    print("row", row)
    print(f"\nRow {row.Index}:")
    print(row.A)
    print(row.B)
    print("Type of row", type(row))
```
