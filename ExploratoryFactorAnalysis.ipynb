Factor Analysis

여러 변수(문항, 지표, 특성 등)들이 서로 밀접하게 상관되어 있을 때
그 변수들을 설명할 수 있는 **공통된 잠재 요인(공통 요인)**을 찾아내는 통계 기법입니다. 
즉, 많은 변수들이 실제로는 몇 개의 요인으로 묶여있을 것이라는 가정 아래, 이를 수학적으로 추출해내는 방법입니다.

EFA, Exploratory Factor Analysis(탐색적 요인 분석)
요인이 몇 개인지, 어떤 구조로 형성되어 있는지 미리 모르는 상태에서 탐색적 기법으로 파악합니다.
결과에 따라 요인의 수를 결정하거나, 측정 도구를 정제하는 데 활용합니다.
