Ranking AI use cases by impact/effort

```python
# Use Case 우선순위 정렬 / Ranking AI use cases by impact/effort
use_cases = [
    {"name": "챗봇 (Chatbot)", "effort": 1, "impact": 3},
    {"name": "추천 시스템 (Recommender)", "effort": 2, "impact": 5},
    {"name": "의료 영상 분석 (Medical Imaging)", "effort": 8 , "impact": 9 },
    {"name": "자율주행 (Autonomy)", "effort": 10, "impact": 10},
]
ranked = sorted(use_cases, key=lambda x: x["impact"]/x["effort"], reverse=True)
for idx, uc in enumerate(ranked, 1):
    print(f"{idx}. {uc['name']} (Efficiency={uc['impact']/uc['effort']:.2f})")
```
