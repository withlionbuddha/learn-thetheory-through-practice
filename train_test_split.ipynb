```python
from sklearn.model_selection import train_test_split
import numpy as np
```

```python
X = np.arange(100).reshape(50, 2) 
y = np.array([0, 1] * 25) # 리스트 [0, 1]를 25번 반복하여 길이가 50인 리스트 생성
```

train_test_split() 

Parameter
  X : input feature data 입력 피처 데이터
  y : label data 레이블(정답) 데이터
  test_size : 테스트 데이터 비율(테스트 데이터 분할 비율)
  train_size : 학습 데이터 비율 (학습 데이터 분할 비율)
  shuffle : 데이터 분할전에 섞을지 여부
  stratify : 데이터의 레이블 분포 비율을 기준으로 학습 데이터와 테스트 데이터의 레이블 분포비율을 동일하게 나눌지 여부

Return
  모든 입력 피처 데이터 다음에 모든 레이블 데이터 순서대로 반환환 
  x_train(학습 입력 피처), x_test(테스트 입력 피처처), y_train(학습 레이블), y_test(테스트 레이블)

```python
X_train, X_test, y_train, y_test = train_test_split(
    X
    , y 
    , test_size=0.2  
    , train_size=0.8
    , random_state=42
    , shuffle=True
    , stratify = y)
```
