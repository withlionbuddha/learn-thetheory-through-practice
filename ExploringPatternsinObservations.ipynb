1. 관측치들의 패턴 탐색 (Exploring Patterns in Observations)  

개념 (Concept):  
관측치 간의 패턴을 분석하여 데이터의 분포, 트렌드, 이상치(outliers)를 확인합니다.  
이 과정에서 데이터의 전체적인 구조를 이해하고, 변수 간의 상관관계나 데이터의 군집화(clustering) 가능성을 탐색할 수 있습니다.  


◾분포 탐색 (Distribution Exploration): 데이터가 특정 분포(정규분포, 균등분포 등)를 따르는지 확인합니다.  
◾시간적 패턴 (Temporal Patterns): 시간에 따른 데이터 변화 추이를 분석합니다.

```python
data = {'날짜': pd.date_range(start='2025-01-01', periods=10)
        , "판매량":[200, 220, 230, 240, 220, 215, 210, 220, 250, 270 ]
        }
df = pd.DataFrame(data)
df
```

```python
plt.plot(df['날짜'], df['판매량'], marker='o')
plt.title("판매량의 시간적 변화(Temporal Change in Sales)")
plt.xlabel("날짜(Date)")
plt.ylabel("판매량(Sales)")

plt.grid()
plt.show()
```

2. 잘못된 자료 탐색 (Identifying Erroneous Data)  

개념 (Concept)  
잘못된 자료는 모델링에 부정적인 영향을 미칠 수 있으므로 탐색해야 합니다.  

◾결측치 탐색 (Missing Values Exploration): 데이터셋에 결측치가 포함되어 있는지 확인합니다.  
◾이상치 탐색 (Outlier Detection): 데이터 범위를 벗어난 값들이 있는지 확인합니다.  
◾불일치 데이터 탐색 (Inconsistent Data Exploration): 데이터 형식이 일관되지 않거나, 정의된 규칙을 위반한 데이터를 찾습니다.

```python
import numpy as np
```

```python
data = {'나이': [25, 30, 35, np.nan, 40, 42, 200],  # 이상치와 결측치 포함
        '점수': [80, 85, 90, 88, np.nan, 92, 95]}
df = pd.DataFrame(data)
df
```

```python
print("결측치 확인(Checking Missing Values) ")
print(df.isnull())
```

```python
print("이상치 탐색 (Outlier Detection)")
print("나이 데이터 중 이상치", df[df['나이'] > 100])
```
