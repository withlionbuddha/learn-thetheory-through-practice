<h1> seaborn.boxplot() </h1>

```python
import matplotlib.pyplot as plt
import numpy as np

# 예시 데이터 생성
data = np.random.normal(0, 1, 100)  # 평균 0, 표준편차 1인 정규분포 데이터를 100개 생성
data = np.concatenate([data, [10, 12, -10]])  # 일부 의도적인 이상치 추가

# Box Plot 그리기
plt.boxplot(data, vert=False)  # 수평 Box Plot으로 그리기
plt.title("Box Plot with Whiskers to Detect Outliers")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 예시 데이터 생성 (정규분포 데이터 100개 생성)
data = np.random.normal(0, 1, 100)

# 일부 의도적인 이상치 추가
data = np.concatenate([data, [10, 12, -10]])  # 10, 12, -10은 이상치로 추가됨

# Box Plot 그리기
plt.boxplot(data, vert=False)  # 수평 Box Plot으로 그리기
plt.title("Box Plot with Whiskers to Detect Outliers")
plt.show()
```
