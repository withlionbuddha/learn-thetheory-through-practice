<h1> boxplot </h1>

목적 <br>
데이터 분포의 중간값과 퍼짐 정도, 이상치 탐지
그룹 간 분포 비교

입력값 <br>
반드시 x의 범주값과 y의 수치형(numeric) 이어야 합니다.


```python
import seaborn as sns
import matplotlib.pyplot as plt
```

<h2> Boxplot showing the Median </h2>

```python
# 예시 데이터 생성
tips = sns.load_dataset("tips")

# Boxplot 그리기
sns.boxplot(x="day", y="total_bill", data=tips)

# 중앙값 표시 (seaborn은 기본적으로 중앙값을 표시함)
plt.title("Boxplot showing the Median")
plt.show()
```

중앙값(median) : 각 박스의 안에 가로로 그려진 선

주어진 데이터셋을 오름차순 또는 내림차순으로 정렬한 후, 가장 중앙에 위치한 값입니다.
홀수 개의 데이터일 경우, 중앙에 있는 값이 중앙값입니다.
짝수 개의 데이터일 경우, 가운데 두 값의 평균을 중앙값으로 계산합니다.

상위 이상치(Outlier) : box의 밖의 상단에 위치한 동그라미 

하위 이상치(Outlier) : box의 밖의 하단에 위치한 동그라미
