```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. 데이터 생성
np.random.seed(42)

# 정상 데이터 (직선 패턴)
x = np.linspace(0, 100, 200)
y = 2 * x + np.random.normal(0, 10, size=x.shape)  # 기울기 2, 약간의 노이즈 추가

# 이상치 추가
outliers_x = np.array([20, 50, 90])
outliers_y = np.array([250, -50, 400])

x = np.concatenate([x, outliers_x])
y = np.concatenate([y, outliers_y])

# 데이터프레임 생성
data = pd.DataFrame({"x": x, "y": y})

# 2. Pairplot 시각화
sns.pairplot(data, kind="scatter")
plt.title("Pairplot with Outliers")
plt.show()

# 3. 산점도에서 이상치 강조
plt.figure(figsize=(8, 6))
plt.scatter(data['x'], data['y'], label="Data", color="blue")
plt.scatter(outliers_x, outliers_y, label="Outliers", color="red", edgecolor="black", s=100)
plt.plot(x, 2 * x, color="green", linestyle="--", label="True Linear Pattern")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Scatter Plot with Outliers")
plt.legend()
plt.grid(True)
plt.show()
```
