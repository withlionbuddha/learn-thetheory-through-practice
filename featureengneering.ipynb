```python
import pandas as pd
import numpy as np
```

```python
# 샘플 시계열 데이터 생성 (10일치)
np.random.seed(42)
dates = pd.date_range("2025-01-01", periods=10, freq="D")
print("dates", dates)

values = np.random.randint(50, 100, size=10)
print("values", values)

df = pd.DataFrame({"date": dates, "value": values})
df.set_index("date", inplace=True)
```

```python
# 3일 이동평균
df["rolling_mean_3d"] = df["value"].rolling(window=3).mean()
df["rolling_mean_3d"]
```

```python
# 2. 교차 예시 (동일 시점 다른 지역 비교)

cross_df = pd.DataFrame({
    "지역A": [100, 120, 130],
    "지역B": [90, 110, 150],
    "지역C": [80, 115, 140]
}, index=["2025-01-01", "2025-01-02", "2025-01-03"])
cross_df
```

```python
# 3. 주파수 (데이터 간격 확인)
freq = pd.infer_freq(df.index)
freq
```
