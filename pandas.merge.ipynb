```python
import pandas as pd
```

<h1> merge </h1>
데이터베이스 스타일의 Join(SQL JOIN) 연산을 수행하여 두 개의 DataFrame(혹은 이름이 지정된 Series)을 병합하는 함수

```python
df_l = pd.DataFrame({
    'key': ['A','B','C'],
    'L':   [1,   2,   3]
})

df_r = pd.DataFrame({
    'key': ['B','C','D'],
    'R':   [4,   5,   6]
})
```

```python
df_l
```

```python
df_r
```

<h1> How = 'inner' </h1>

공통된 key를 기준으로 dataframe or series를 병합합니다다. 

```python
pd.merge(df_l, df_r, on='key', how='inner')
```

<h1> How = 'left' </h1>

```python
pd.merge(df_l, df_r, on='key', how='left')
```

<h1> How = 'right' </h1>

```python
pd.merge(df_l, df_r, on='key', how='right')
```

<h1> How = 'outer' </h1>

```python
pd.merge(df_l, df_r, on='key', how='outer')
```
