```python
L = ['a', 'b', 'c', 'd', 'e']
```

```python
L[0]    # 'a'    – 첫 번째 요소
```

```python
L[2]    # 'c'    – 세 번째 요소
```

```python
L[-1]   # 'e'    – 마지막 요소
```

```python
L[-2]   # 'd'    – 뒤에서 두 번째 요소
```

```python
L = [0,1,2,3,4,5,6]
```

```python
L[1:4]     # [1,2,3]    – 인덱스 1~3 (4 미포함)
```

```python
L[:3]      # [0,1,2]    – 시작 생략 → 0부터
```

```python
L[4:]      # [4,5,6]    – 끝 생략 → 끝까지
```

```python
L[:]       # [0,1,2,3,4,5,6] – 전체 복사
```

```python
L = [10,20,30,40,50,60,70]

```

```python
L[::2]     # [10,30,50,70]  – 2칸씩 건너뛰며
```

```python
L[1::2]    # [20,40,60]     
```

```python
L[::-1]    # [70,60,...,10] – 역순
```

```python
L[5:1:-2]  # [60,40]        – 5→2 인덱스, 2칸씩
```

```python
L = [5, 10, 15, 20, 25]
```

```python

# 값이 15 이상인 요소만
[x for x in L if x >= 15]            # [15, 20, 25]

```

```python
# 인덱스와 함께
[i for i, x in enumerate(L) if x % 20 == 0]  # [3]


```

```python
# 값에 +1 적용
[x+1 for x in L]                     # [6,11,16,21,26]
```

```python
L = [2, 4, 6, 7, 8]
```

```python

# 조건에 맞는 첫 번째 값의 인덱스
idx = next((i for i, x in enumerate(L) if x % 2 != 0), None)
# idx == 3 (값 7)

```

```python

# filter 로 조건에 맞는 값만
list(filter(lambda x: x % 2 == 0, L))  # [2,4,6,8]
```
