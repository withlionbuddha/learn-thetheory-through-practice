```python
# 1. 인공 데이터 생성 (예: 300명의 응답, 10개 문항)
np.random.seed(42)
data = pd.DataFrame(np.random.randint(1, 6, size=(300, 10)), 
                    columns=[f"Q{i+1}" for i in range(10)])
print("데이터 미리보기:")
print(data.head())
```

```python
# 2. Bartlett의 구형성 검정 수행
chi_square_value, p_value = calculate_bartlett_sphericity(data)
print("\nBartlett's Test 결과:")
print("Chi-square =", chi_square_value)
print("p-value =", p_value)
# p-value가 작으면 변수들 간에 상관관계가 존재함을 의미합니다.
```

```python



# 3. KMO 검정 수행
kmo_all, kmo_model = calculate_kmo(data)
print("\nKMO 검정 결과:")
print("각 변수에 대한 KMO 값:", kmo_all)
print("전체 KMO 값:", kmo_model)
# 전체 KMO 값이 0.6 이상이면 요인 분석 진행이 적합합니다.

# 4. 변수 간 상관관계 시각화 (Pairplot)
plt.figure(figsize=(10, 10))
sns.pairplot(data)
plt.suptitle("변수들 간의 관계 (Pairplot)", y=1.02)
plt.show()
```
