```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 군집에서의 이상치 예시
np.random.seed(42)

# 군집 데이터 생성
cluster_1_height = np.random.normal(160, 5, 100)  # 첫 번째 군집 (평균 160cm)
cluster_1_weight = np.random.normal(50, 5, 100)  # 첫 번째 군집 (평균 50kg)

cluster_2_height = np.random.normal(175, 5, 100)  # 두 번째 군집 (평균 175cm)
cluster_2_weight = np.random.normal(70, 5, 100)  # 두 번째 군집 (평균 70kg)

# 군집 내 이상치 추가
outliers_height = np.array([162, 176, 180])
outliers_weight = np.array([90, 95, 100])  # 군집 내 이상치

# 데이터 결합
height_with_clusters = np.concatenate([cluster_1_height, cluster_2_height, outliers_height])
weight_with_clusters = np.concatenate([cluster_1_weight, cluster_2_weight, outliers_weight])

# 데이터프레임 생성
cluster_data = pd.DataFrame({"Height": height_with_clusters, "Weight": weight_with_clusters})

# Pairplot 시각화
sns.pairplot(cluster_data, kind="scatter")
plt.title("Pairplot of Height and Weight with Cluster Outliers")
plt.show()

# 산점도에서 군집 이상치 강조
plt.figure(figsize=(8, 6))
plt.scatter(cluster_1_height, cluster_1_weight, label="Cluster 1", color="blue")
plt.scatter(cluster_2_height, cluster_2_weight, label="Cluster 2", color="green")
plt.scatter(outliers_height, outliers_weight, label="Outliers", color="red", edgecolor="black", s=100)

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Height vs Weight with Cluster Outliers")
plt.legend()
plt.grid(True)
plt.show()

# 그래프 특징 설명:
# 이 그래프는 두 개의 뚜렷한 군집(Cluster 1, Cluster 2)과 군집 내에서 발생한 이상치를 보여줍니다.
# 이상치는 군집의 중심에서 멀리 떨어진 데이터로, 붉은색 점으로 강조되었습니다. 
# 군집 간 데이터 분포와 이상치를 비교하여 데이터의 품질을 평가할 수 있습니다.
```
