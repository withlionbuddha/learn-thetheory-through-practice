<h1> CountPlot </h1>

목적

각 범주별 데이터 개수 확인 <br>
    범주형 변수의 각 범주에 속하는 데이터의 수를 쉽게 확인할 수 있습니다. 이는 데이터의 분포를 한눈에 파악할 수 있는 좋은 방법입니다.

클래스 불균형 탐색 <br>
    데이터에서 클래스 불균형을 시각적으로 확인하는 데 유용합니다. 예를 들어, 한 클래스가 다른 클래스에 비해 지나치게 많거나 적은 경우 이를 시각적으로 쉽게 파악할 수 있습니다.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python

# 예시 데이터 생성
data = {
    'Category': ['A', 'B', 'A', 'C', 'B', 'A', 'A', 'C', 'B', 'A', 'C', 'C', 'B', 'A', 'C']
}

df = pd.DataFrame(data)
```

```python
# countplot을 사용하여 범주형 변수의 빈도 시각화
plt.figure(figsize=(8, 6))
sns.countplot(x='Category', data=df)
plt.title('Category Count')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```
