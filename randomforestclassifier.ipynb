랜덤포레스트(Random Forest)는 결정의사트리(Decision Tree)를 기본 모델로 사용하는 배깅(Bagging) 방식의 앙상블 모델 입니다.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import plot_tree
```

```python
# 데이터 생성
X, y = make_classification(n_samples=100,        #100개의 데이터(샘플플)
                           n_features=10,        # 10개의 피처
                           n_informative=4,      # 4개의 실제 타깃(클래스)를 결정하는 데 기여하는 유익한(informative) 피처의 수
                           n_redundant=2,        # 2개의 중복 피처
                           n_classes=2,          # 이진 분류
                           n_clusters_per_class=1,
                           random_state=42)
```

RandomForestClassifier는 각 결정 트리의 분류 결과(클래스)를 다수결(voting) 방식으로 결합하여 최종 클래스를 결정합니다.

n_estimators=100  100개의 결정 트리를 생성합니다. 
max_depth=5 각 트리의 깊이를 5로 제한합니다.
min_samples_split=30 노드를 분할하기 위한 최소 샘플 수를 30으로 설정합니다.
min_samples_leaf=15 리프 노드에 반드시 존재해야 하는 최소 샘플 수를 설정합니다.
random_state=42 행할 때마다 동일한 결과를 얻도록 난수 생성의 초기값을 고정합니다.

```python
# 모델 학습
model = RandomForestClassifier(n_estimators=100, max_depth=5, min_samples_split=30, min_samples_leaf=15, random_state=42)
model.fit(X, y)
```

```python
# 첫 번째 트리 시각화
estimator = model.estimators_[0]

plt.figure(figsize=(12, 8))
plot_tree(estimator, filled=True)
plt.show()
```

```python
class_y = model.predict(X)
print("classifiyer = " , class_y)
print("label = ", y)
```
