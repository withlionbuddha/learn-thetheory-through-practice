데이터 정규화(예: MinMax 스케일링) 과정에서:

빼기(-)와 더하기(+) 연산은 데이터를 좌표상에서 이동(shift)시키는 역할을 합니다.
예를 들어, 데이터를 0부터 시작하도록 맞추기 위해 각 값에서 데이터셋의 최소값을 빼고, 필요하면 원하는 시작점을 맞추기 위해 일정 값을 더합니다.

나누기(/)와 곱하기(*) 연산은 데이터를 원하는 범위로 조정하기 위해 크기를 축소하거나 확대하는(스케일링) 역할을 합니다.

이러한 연산을 통해 데이터의 상대적 비율은 유지하면서, 전체 데이터가 지정한 범위(예: 0~1)에 들어오도록 변환됩니다.

data 중 스케일된 x값을 구하는 공식
x_scaled = feature_range_min +  (x - data_min) x  ( (feature_range_max − feature_range_min) / (data_max − data_min) )

```python
import numpy as np
import pandas as pd

# 원본 데이터 준비
data = np.array([10, 20, 30, 40])
df = pd.DataFrame(data, columns=['Value'])
```

```python
class min_max_normalizer :
    def __init__(self, data):
        self.data = data

    def set_feature_range(self, feature_range) :
        self.feature_range = feature_range

    def normalize(self) :
        
        # 1 데이터에서 최소값을 빼서 shift를 적용하는 단계
        shifted_data = data - data.min() 
        # 2  scale 계수를 곱해 비율에 맞게 크기를 scaling
        scaled_data = shifted_data * self._get_scale_coef()
        # 3 데이터를 feature_range의 범위내로 맞추는 단계
        normalized_data = self.feature_range.min() + scaled_data
        return normalized_data

    def _get_scale_coef(self) :
        scale_coef = (self.feature_range.max() - self.feature_range.min()) / (self.data.max() - self.data.min())
        return scale_coef
```

```python
feature_range = np.array([0,1])

normalizer = min_max_normalizer(data)
normalizer.set_feature_range(feature_range)
print(normalizer.normalize())
```

```python
from sklearn.preprocessing import MinMaxScaler

# MinMaxScaler 적용
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_values  = scaler.fit_transform(df[['Value']])

df['Scaled'] = scaled_values
print("\n정규화된 데이터:")
print(df)

# 5. 내부 계산된 파라미터 확인
print("\n계산된 내부 파라미터:")
print("data_min_:", scaler.data_min_)
print("data_max_:", scaler.data_max_)
print("scale_:", scaler.scale_)
```
