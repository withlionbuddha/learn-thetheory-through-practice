```python
import numpy as np
import pandas as pd
```

결측치의 표준값
    Numeric 열:
    np.nan 값이 포함되어 있어 수치형 데이터의 결측치를 표현합니다.

    Object 열:
    None 값이 포함되어 있어 문자열 등 객체 타입에서의 결측치를 나타냅니다.

    Datetime 열:
    pd.NaT 값이 포함되어 있어 datetime 타입의 결측치를 나타냅니다.

    Nullable 열:
    pd.NA 값이 포함되어 있으며, 이는 pandas에서 제공하는 Nullable 데이터 타입의 결측치 표준값입니다.

```python
# 예제 데이터 생성
data = {
    'Numeric': [1, np.nan, 3, 4],                      # np.nan 포함
    'Object': [None, 'b', 'c', 'd'],                    # None 포함
    'Datetime': [pd.NaT, pd.Timestamp('2020-01-01'), pd.NaT, pd.Timestamp('2020-06-15')],  # pd.NaT 포함
    'Nullable': [pd.NA, pd.NA, 'x', 'y']               # pd.NA 포함 (pandas의 Nullable 타입)
}

df = pd.DataFrame(data)

print("DataFrame:")
print(df)
```

pandas.isna()
pandas.isnull()
    결측치(np.nan, None, pd.NaT, pd.NA 등)를 판별하기 위해 사용되며, isna()와 isnull() 같은 기능을 수행합니다.

```python
print("\n pd.isna(df):")
print(pd.isna(df))
```
