```python
import pandas as pd
import numpy as np
```

dataframe.drop() 
    원본 DataFrame을 변경하지 않고, 지정한 열이나 행을 삭제한 새 DataFrame 객체를 반환합니다.
    inplace=True 옵션설정시 원본 DataFrame 직접 수정

```python
# DataFrame 생성
data = {'A': [1, 2, 3], 'B': [4, 5, 6], 'class': [7, 8, 9]}
df = pd.DataFrame(data)

# drop() 호출하면 새로운 dataframe이 생성된다.
new_df = df.drop(columns=['class'])

print("원본 DataFrame:")
print(df)  # 'class' 열이 그대로 있음

print("\n새로운 DataFrame:")
print(new_df)  # 'class' 열이 삭제된 새로운 DataFrame
```

dataframe.drop()

    용도 
        인덱스(행)나 컬럼(열)의 라벨(label)을 기준으로 지정된 행이나 열을 삭제할 때 사용합니다.
        라벨에 의한 명시적으로 데이터를 제거할때 사용합니다.

```python
# 인덱스(행) 삭제
df.drop([0, 2], axis=0, inplace=True)

# 컬럼(열) 삭제
df.drop(['A', 'B'], axis=1, inplace=True)
```

불리언 인덱싱(Boolean Indexing)

dataframe.dropna() 

    DataFrame에서 결측치(NA, NaN 등)가 포함된 행 또는 열을 제거하는 데 사용됩니다.  
    기본적으로 결측치가 하나라도 포함된 행을 삭제하지만, 옵션을 통해 열을 삭제하거나 특정 조건을 설정할 수 있습니다.

    axis
        axis=0 (기본값)은 결측치가 있는 행을 제거합니다.
        axis=1은 결측치가 있는 열을 제거합니다.
    how
        'any' (기본값)은 한 개라도 결측치가 있으면 삭제합니다.
        'all' 'all'은 모든 값이 결측치인 행(또는 열)만 삭제합니다.
    thresh
        지정된 axis에 대해서 결측치가 아닌 값의 최소 개수를 지정합니다.
    subset
        특정 열에 대해서만 결측치 여부를 확인하여 삭제할 수 있습니다.
    inplace
        inplace=True로 설정하면 원본 DataFrame을 직접 수정합니다.

```python
df_na = df.dropna(axis=0, how='any')
df_na
```

dataframe.loc()

```python
df = pd.DataFrame(
    {'age':[25,30,22,40],'city':['Seoul','Gyeonggi','Gyeongsangbuk','Gwangju']},
    index=['ROH','KIM','CHO', 'LEE']
)
df
```

```python
# 1) 단일 라벨 행
df.loc['KIM']                  # Series 반환
```

```python
# 2) 여러 행 & 여러 열 선택
df.loc[['KIM','ROH'], ['age','city']]
```

```python
# 3) 라벨 슬라이스 (포함 범위)
df.loc['KIM':'CHO', :]        # 세 행 모두
```

```python
# 4) 불리언 마스크
df.loc[df['age'] > 24, 'city']
```

```python
# 5) 값 수정
df.loc['ROH', 'city'] = 'Gwangju'
```
